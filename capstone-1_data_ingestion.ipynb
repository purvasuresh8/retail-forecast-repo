```python
# =========================
# Part 1 — Setup
# =========================
!pip install --quiet gdown openpyxl

import os
import re
import pandas as pd
from IPython.display import display

try:
    from google.colab import drive
except Exception:
    drive = None

# =========================
# Part 2 — Mount Drive
# =========================
mount_drive = True

if mount_drive and drive is not None:
    drive.mount('/content/drive', force_remount=False)
    DRIVE_BASE = '/content/drive/MyDrive'
else:
    DRIVE_BASE = '/content'  )


os.makedirs(DRIVE_BASE, exist_ok=True)
print("Save folder:", DRIVE_BASE)

# =========================
# Part 3 — Download dataset using gdown
# =========================
DRIVE_LINK = "https://docs.google.com/spreadsheets/d/1iGfXYkIyRrwdZUos6EaQdCV8d3ZxWFYzbxQPi1qoGag/edit?usp=sharing"

def get_drive_id(url: str) -> str:
    patterns = [r"/d/([a-zA-Z0-9_-]+)", r"id=([a-zA-Z0-9_-]+)"]
    for p in patterns:
        m = re.search(p, url)
        if m:
            return m.group(1)
    raise ValueError("Couldn't extract file id from the Drive link. Please paste the share link.")

file_id = get_drive_id(DRIVE_LINK)
download_url = f"https://drive.google.com/uc?id={file_id}"
local_path = os.path.join(DRIVE_BASE, "Retail_Data_Set.csv")

print("Downloading to:", local_path)
import gdown
gdown.download(download_url, local_path, quiet=False)

# =========================
# Part 4 — Load dataset into pandas
# =========================
print("Loading dataset from:", local_path)

try:
    df = pd.read_csv(local_path, encoding="latin1")
except Exception:
    try:
        df = pd.read_excel(local_path, engine="openpyxl")
    except Exception as e:
        raise RuntimeError("Cannot read file; check format.") from e

print("Shape:", df.shape)
display(df.head())

# =========================
# Part 5 — Inspect dataset
# =========================
print("\n--- Data Types ---")
display(df.dtypes)

print("\n--- Summary Statistics ---")
display(df.describe(include="all"))

print("\n--- Missing Values per Column ---")
display(df.isna().sum())

# =========================
# Part 6 — Remove duplicates
# =========================
before = df.shape[0]
df = df.drop_duplicates()
after = df.shape[0]

print(f"Removed {before - after} duplicate rows")

# =========================
# Part 7 — Save the raw CSV into the chosen folder
# =========================
# CSV ready to import into pgAdmin as retail.raw_sales
save_path = os.path.join(DRIVE_BASE, "raw_sales.csv")
df.to_csv(save_path, index=False)

print("Saved raw dataset to:", save_path)

# =========================
# Part 8 — Quick summary message
# =========================
print("Notebook 1 done. You can now import 'raw_sales.csv' into pgAdmin under 'retail.raw_sales'.")
```
